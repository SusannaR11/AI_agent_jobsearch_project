## EDA on jobads api data to duckdb

```python
import requests

url = "https://jobsearch.api.jobtechdev.se/search"
params= {"q": "Flen", "limit":5}

r = requests.get(url, params=params, timeout=30)
r.raise_for_status()
data= r.json()

print(data.keys())
print("hits:", len(data["hits"]))
print(data["hits"][0].keys())
```

```python
from collections import Counter

hits = data["hits"]

c = Counter()
for h in hits:
    og = h.get("occupation_group") or {}
    label = og.get("label")
    if label:
        c[label] += 1

job_count = c.most_common()
print(job_count[:10])
```
